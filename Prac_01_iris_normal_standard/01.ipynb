# Step 1–2: Load dataset & add column names

```python
import pandas as pd

# Load dataset (if CSV without headers)
df = pd.read_csv("iris.csv")   # remove header=None

# Add column names
df.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species']

# Display dataset
print(df.head())
```

# Step 3: First & Last 5 rows

```python
print("First 5 rows:\n", df.head())
print("\nLast 5 rows:\n", df.tail())
```

# Step 4: Initial Statistics

```python
print(df.describe())
```

# Step 5: Variable Description

```python
print(df.info())
```

# Step 6: Check & Convert Data Types

```python
print(df.dtypes)

# Example conversion (if needed)
df['species'] = df['species'].astype('category')
```

# Step 7: Normalization & Standardization




```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df_norm = df.copy()

df_norm[['sepal_length','sepal_width','petal_length','petal_width']] = scaler.fit_transform(
    df[['sepal_length','sepal_width','petal_length','petal_width']]
)

print(df_norm.head())
```

# Step 8: Label Encoding (Species → Numeric)

```python
from sklearn.preprocessing import LabelEncoder

# Create encoder object
le = LabelEncoder()

# Apply on 'species' column
df['species_encoded'] = le.fit_transform(df['species'])

# Display result
print(df[['species', 'species_encoded']].head())
```
