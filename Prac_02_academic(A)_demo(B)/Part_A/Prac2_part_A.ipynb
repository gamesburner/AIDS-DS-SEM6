# Part A: Handling Missing Values

## Step 1: Load dataset

```python
import pandas as pd

df = pd.read_csv("academic.csv")  
print(df.head())
```

## Step 2: Check missing values

```python
print("Original Data:\n")
print(df)

print("\nMissing values:\n")
print(df.isnull().sum())
```

## Step 3: Fill missing values

### Using Mean

```python
df_mean = df.copy()
df_mean.fillna(df_mean.mean(numeric_only=True), inplace=True)
print("\nAfter Mean Filling:\n")
print(df_mean)
```

### Using Median

```python
df_median = df.copy()
df_median.fillna(df_median.median(numeric_only=True), inplace=True)
print("\nAfter Median Filling:\n")
print(df_median)
```

### Using Standard Deviation

```python
df_std = df.copy()

for col in df_std.select_dtypes(include=['float64', 'int64']).columns:
    df_std[col] = df_std[col].fillna(df_std[col].std())

print("\nAfter Std Dev Filling:\n")
print(df_std)
```
