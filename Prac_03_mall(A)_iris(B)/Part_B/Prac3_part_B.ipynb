# Part B: Iris Dataset Statistics

## Step 1–2: Load + Add Column Names

```python
import pandas as pd

df2 = pd.read_csv("Iris.csv", header=None)

df2.columns = ['Id','sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species']
```

```python
df2 = df2.drop(columns=["Id"])
```

## Step 3: Statistical Details for Each Species


```python
for species, data in df2.groupby('species'):
    print(f"\nStatistics for {species}:\n")
    print(data.describe())
```
