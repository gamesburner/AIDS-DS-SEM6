# Part A: Mall Customers Dataset

## Step 1: Load dataset

```python
import pandas as pd

df = pd.read_csv("Mall_Customers.csv")
print(df.head())
```

## Step 2: Mean, Median, Mode, Std

```python
print("Mean:\n", df.mean(numeric_only=True))
print("\nMedian:\n", df.median(numeric_only=True))
print("\nMode:\n", df.mode())
print("\nStandard Deviation:\n", df.std(numeric_only=True))
```

## Step 3: Create Age Groups

```python
bins = [0, 20, 40, 60, 100]
labels = ['Teen', 'Young', 'Adult', 'Senior']

df['Age_Group'] = pd.cut(df['Age'], bins=bins, labels=labels)
```

## Step 4: Summary Statistics (Income grouped by Age)

```python
group_stats = df.groupby('Age_Group', observed=True)['Annual Income (k$)'].describe()
print(group_stats)
```

## Step 5: Convert Age Group to Numeric (List)


```python
age_group_numeric = df['Age_Group'].cat.codes
print(age_group_numeric.head())
```
