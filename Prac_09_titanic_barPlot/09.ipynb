```python
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset('titanic')
print(df.head())
```

# 1. Bar Plot → Survival vs Passenger Class

```python
sns.countplot(x='class', hue='survived', data=df)
plt.title("Survival Count by Passenger Class")
plt.show()
```

# 2. Histogram → Age (Survivors vs Non-survivors)

```python
sns.histplot(data=df, x='age', hue='survived', bins=30, kde=True)
plt.title("Age Distribution (Survived vs Not)")
plt.show()
```

# 3. Box Plot → Age vs Gender & Survival

```python
sns.boxplot(x='sex', y='age', hue='survived', data=df)
plt.title("Age Distribution by Gender & Survival")
plt.show()
```

# 5. Correlation Heatmap

```python
numeric_df = df.select_dtypes(include=['int64','float64'])

sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```

# 6. Bar Plot → Survival by Embarked Port

```python
sns.countplot(x='embarked', hue='survived', data=df)
plt.title("Survival by Embarkation Port")
plt.show()
```
